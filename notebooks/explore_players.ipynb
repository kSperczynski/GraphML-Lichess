# Explore players

```python
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
```

```python
PATH_PLAYERS_FILE = Path("..", "data", "player_features.csv")
df = pd.read_csv(filepath_or_buffer=PATH_PLAYERS_FILE)
display(df.nunique())
```

```python
df.hist(bins=30, figsize=(15, 10), edgecolor='black')
plt.suptitle('Distribution of Player Statistics', fontsize=16)
plt.tight_layout() # Prevents subplot titles and labels from overlapping
plt.show()
```
