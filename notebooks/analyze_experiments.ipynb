# Saved experiment analysis
Read existing artifacts only. No training is started.
Set the run roots and a NEW output directory before executing.

```python
from pathlib import Path
from graphml_lichess.analysis import ExperimentCollection
from graphml_lichess.analysis.plots import plot_summary

RUN_ROOTS = [Path("../runs/training")]
OUTPUT = Path("../runs/analysis/notebook_first")
experiments = ExperimentCollection.from_roots(RUN_ROOTS)
```

```python
summary, per_split = experiments.summarize(unit="training")
display(summary[["model_type", "config_id", "split_id", "n_units", "rmse_mean", "rmse_std"]])
display(experiments.diagnostics)
```

```python
destination = experiments.export(OUTPUT, unit="training")
plot_summary(summary, destination, metric="rmse", phase="test", interval="sd")
```

## Multiple player splits
Use `summarize(unit="split")` only for genuinely different split fingerprints.
Training seeds are averaged within each split first. CI plots require at least three splits.
Overlapping random splits are not necessarily independent; t intervals are descriptive.
Legacy runs remain unverified if historical input hashes were not recorded.
