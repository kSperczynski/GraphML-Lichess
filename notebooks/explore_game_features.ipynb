# Explore players

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
PATH_PLAYERS_FILE = Path("..", "data", "game_features.csv")
df = pd.read_csv(filepath_or_buffer=PATH_PLAYERS_FILE)
display(df.nunique())
```

```python
numerical_df = df.select_dtypes(include=['number'])
numerical_df = numerical_df.replace([np.inf, -np.inf], np.nan)
numerical_df.hist(bins=30, figsize=(20, 15), edgecolor='black')

plt.suptitle('Distribution of Node Statistics', fontsize=16)
plt.tight_layout()
plt.show()
```
